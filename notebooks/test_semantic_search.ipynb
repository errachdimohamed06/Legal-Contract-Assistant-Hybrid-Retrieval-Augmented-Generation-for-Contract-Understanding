```python
import sys
import yaml
from pathlib import Path

sys.path.append(".")

import chromadb
from langchain_chroma import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings

chroma_folder = Path(r"C:\Users\Errachdi\OneDrive - Institut National de Statistique et d'Economie Appliquee\Bureau\projet saad\legal-rag-assistant\outputs\chroma_db")

with open("../configs/model_config.yaml", "r") as f:
    model_config = yaml.safe_load(f)

embedding_model = HuggingFaceEmbeddings(
    model_name=model_config["embedding"]["model_name"],
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": True}
)

client = chromadb.PersistentClient(path=str(chroma_folder))
vector_store = Chroma(
    client=client,
    collection_name="cuad_contracts",
    embedding_function=embedding_model
)

print(f"Total chunks in DB: {vector_store._collection.count()}")

def test_query(query_text, k=5):
    print(f"\n{'='*80}")
    print(f"QUERY: {query_text}")
    print(f"{'='*80}")
    
    results = vector_store.similarity_search_with_score(query_text, k=k)
    
    if not results:
        print("❌ No results found.")
        return
    
    for i, (doc, score) in enumerate(results):
        print(f"\nRank {i+1} (Score: {score:.4f} - lower is better)")
        print(f"  File    : {doc.metadata.get('source', 'Unknown')}")
        print(f"  Header  : {doc.metadata.get('header', 'Unknown')}")
        snippet = doc.page_content.replace('\n', ' ').strip()
        print(f"  Snippet : {snippet[:300]}...")
        print("-" * 60)

queries = [
    "Governing Law",
    "What is the governing law of this contract?",
    "Termination for convenience",
    "Expiration Date of the agreement",
    "Confidentiality clause",
    "Indemnification obligations",
    "Insurance requirements",
    "Intellectual Property ownership",
    "License Grant",
    "Non-Compete agreement",
    "Exclusivity clause",
    "Force Majeure event",
    "Audit Rights",
    "Payment terms",
    "How much does it cost?",
    "Assignment of rights",
    "Notices provision"
]

for q in queries:
    test_query(q, k=3)
```
